```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace', 'color':'green', 'size':14}

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
        marker = '*',
        markersize = 12, 
        color = 'indigo',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')

ax.grid(axis = 'y')
```

```python
# Playing around with code at end of class

np.random.seed(101)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
        alpha = 0.3,
        marker = "o",
        markersize = 2,
        color = 'goldenrod',
        linestyle = '-',
        linewidth = 1,
        markeredgecolor = 'mediumturquoise',
        markerfacecolor = 'darkslateblue')

ax.grid(axis = 'x')

ax.set_title('Annual harvest of grapes over time')
ax.set_ylabel('Annual harvest of grapes (Pints)')
ax.set_xlabel('Years since 2000')
```
